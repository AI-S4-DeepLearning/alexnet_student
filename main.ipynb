# Stap 4 - Gridsearch

**Opdracht:** Voer gridsearch uit door de juiste parameterwaarden te kiezen

```python
param_grid = {
    'batch_size': [],
    'learning_rate': [],
    'dropout_rate': []
}
```

```python
def grid_search(train_ds, val_ds, epochs=20):
    max_val_acc = 0
    max_trainer = None

    for lr in param_grid['learning_rate']:
        for bs in param_grid['batch_size']:
            for dropout in param_grid['dropout_rate']:
                print(f"\n{'='*100}")
                print(f"Training: learning rate={lr}, batch size={bs}, dropout={dropout}")
                print(f"{'='*100}")
            
                trainer = AlexNetTrainer(num_classes=len(generator.classes), 
                                        learning_rate=lr,
                                        batch_size=bs,
                                        dropout_rate=dropout,
                                        device=device)
                
                trainer.fit(train_ds, val_ds, bs, epochs)
                
                if trainer.history['val_acc'][-1] > max_val_acc:
                    max_val_acc = trainer.history['val_acc'][-1]
                    max_trainer = trainer
        
    return max_trainer
```

```python
max_trainer_unbalanced = grid_search(train_dataset, val_dataset)
```
